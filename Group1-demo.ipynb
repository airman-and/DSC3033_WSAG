## Limited Visual Inference

The inference cell uses the official AGD20K-Seen checkpoint and saves only five examples. Subprocess output is captured so the notebook does not fill with progress bars. On this server, saved demo visuals are reused by default. In Colab, the cell runs inference after the checkpoint and data are ready.

```python
#OWN CODE: Run limited visual inference or reuse saved outputs for a fast live demo.
existing_visuals = (DEMO_OUTPUT_DIR / 'overlay').exists() and any((DEMO_OUTPUT_DIR / 'overlay').glob('*.png'))
fallback_visuals = (FALLBACK_VISUAL_DIR / 'overlay').exists() and any((FALLBACK_VISUAL_DIR / 'overlay').glob('*.png'))
RUN_VISUAL_INFERENCE = IS_COLAB or not (existing_visuals or fallback_visuals)

if RUN_VISUAL_INFERENCE:
    DEMO_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    command = [
        sys.executable, 'test.py',
        '--data_root', str(DATA_ROOT),
        '--model_file', str(CHECKPOINT),
        '--divide', 'Seen',
        '--gpu', '0',
        '--test_batch_size', '1',
        '--test_num_workers', '0',
        '--save_visuals',
        '--save_heatmaps',
        '--max_save_images', '5',
        '--save_path', str(DEMO_OUTPUT_DIR),
    ]
    print('Running:', ' '.join(command))
    result = subprocess.run(command, text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, check=True)
    metric_lines = [line for line in result.stdout.splitlines() if line.startswith(('KLD,', 'reeKLD,', 'remKLD,'))]
    print('\n'.join(metric_lines[-3:]))
else:
    print('Visual inference is skipped because saved demo visuals are available.')
    print('Set RUN_VISUAL_INFERENCE = True manually to regenerate five examples.')
```

```python
#OWN CODE: Select the current visual output directory or a saved fallback directory.
def choose_visual_dir():
    if (DEMO_OUTPUT_DIR / 'overlay').exists() and list((DEMO_OUTPUT_DIR / 'overlay').glob('*.png')):
        return DEMO_OUTPUT_DIR
    if (FALLBACK_VISUAL_DIR / 'overlay').exists() and list((FALLBACK_VISUAL_DIR / 'overlay').glob('*.png')):
        return FALLBACK_VISUAL_DIR
    raise FileNotFoundError('No visual outputs are available. Run the visual inference cell first.')

VISUAL_DIR = choose_visual_dir()
print('Using visual directory:', VISUAL_DIR)
```

```python
#OWN CODE: Build a lookup from saved visualization filenames to original images and GT masks.
from PIL import Image
from IPython.display import display

TEST_IMAGE_ROOT = DATA_ROOT / 'Seen/testset/egocentric'
MASK_ROOT = DATA_ROOT / 'Seen/testset/GT'
PREDICTION_SUFFIXES = ['ego_pred', 'refined_ego', 'refined_mean']

def find_image_for_mask(mask_path):
    relative_stem = mask_path.relative_to(MASK_ROOT).with_suffix('')
    for extension in ['.jpg', '.jpeg', '.png']:
        candidate = TEST_IMAGE_ROOT / relative_stem.with_suffix(extension)
        if candidate.exists():
            return candidate
    return None

def build_visual_lookup():
    lookup = {}
    for mask_path in MASK_ROOT.rglob('*.png'):
        stem = '_'.join(mask_path.relative_to(MASK_ROOT).with_suffix('').parts)
        image_path = find_image_for_mask(mask_path)
        if image_path is None:
            continue
        for suffix in PREDICTION_SUFFIXES:
            lookup[f'{stem}_{suffix}.png'] = {'image': image_path, 'mask': mask_path, 'prediction': suffix}
    return lookup

visual_lookup = build_visual_lookup()
print('Lookup entries:', len(visual_lookup))
```

```python
#OWN CODE: Summarize official checkpoint metrics without printing the full test log.
metric_rows = [
    {'Split': 'Seen', 'Prediction': 'ego_pred', 'KLD': 1.142, 'SIM': 0.415, 'NSS': 1.303},
    {'Split': 'Seen', 'Prediction': 'refined ego-ego', 'KLD': 1.124, 'SIM': 0.433, 'NSS': 1.280},
    {'Split': 'Seen', 'Prediction': 'refined ego-mean', 'KLD': 1.140, 'SIM': 0.422, 'NSS': 1.283},
]

display_table(metric_rows, ['Split', 'Prediction', 'KLD', 'SIM', 'NSS'])
print('The refined ego-ego map is the result used for the main live-demo claim.')
```

## Final Inference Visualization

The final code cell shows the pretrained-checkpoint inference result. Each row uses a real AGD20K image and its affordance label text, then displays the original image, ground-truth mask, predicted heatmap, and overlay.

```python
#OWN CODE: Display five refined predictions as original image, GT mask, heatmap, and overlay.
def resize_for_display(image, size=(224, 224)):
    return image.convert('RGB').resize(size)

def make_panel(image_path, mask_path, heatmap_path, overlay_path):
    images = [
        resize_for_display(Image.open(image_path)),
        resize_for_display(Image.open(mask_path)),
        resize_for_display(Image.open(heatmap_path)),
        resize_for_display(Image.open(overlay_path)),
    ]
    width, height = images[0].size
    panel = Image.new('RGB', (width * len(images), height), color='white')
    for index, image in enumerate(images):
        panel.paste(image, (index * width, 0))
    return panel

overlay_files = sorted((VISUAL_DIR / 'overlay').glob('*_refined_ego.png'))[:5]
for overlay_path in overlay_files:
    item = visual_lookup.get(overlay_path.name)
    if item is None:
        print('Skipping unmatched visualization:', overlay_path.name)
        continue
    heatmap_path = VISUAL_DIR / 'heatmap' / overlay_path.name
    rel_parts = item['mask'].relative_to(MASK_ROOT).parts
    affordance_text = rel_parts[0] if rel_parts else 'unknown'
    object_name = rel_parts[1] if len(rel_parts) > 1 else 'unknown'
    print(overlay_path.name)
    print('Affordance label text:', affordance_text, '| Object:', object_name)
    display(make_panel(item['image'], item['mask'], heatmap_path, overlay_path))
    print('Panel order: original image | ground-truth mask | predicted heatmap | overlay')
```
